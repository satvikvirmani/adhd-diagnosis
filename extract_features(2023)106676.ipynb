```python
import os
import numpy as np
import scipy.io as sio
import matlab.engine
import pandas as pd
```

```python
from helpers.python.feature_extraction import extract_from_signals
```

```python
eng = matlab.engine.start_matlab()
eng.addpath(r'/Users/satvik/Documents/projects/adhd/helpers/matlab', nargout=0)
print("MATLAB started. Test sqrt(16):", eng.sqrt(16.0))
```

```python
data_dir = "data/ADHD_part1"
filename = "v1p.mat"
```

```python
filepath = os.path.join(data_dir, filename)

mat = sio.loadmat(filepath)
signal = mat[list(mat.keys())[-1]]

print(f"\nProcessing: {filepath} with shape: {signal.shape}")
```

```python
# ------------------ EMD ------------------
emd_signals = eng.empirical_mode_decomposition(filepath, 1, 6)
emd_signals_np = np.array(emd_signals)
print("EMD shape:", emd_signals_np.shape)

# ------------------ EWT ------------------
ewt_signals = eng.empirical_wavelet_transformation(filepath, 1, 6)
ewt_signals_np = np.array(ewt_signals)
print("EWT shape:", ewt_signals_np.shape)

# ------------------ VMD ------------------
vmd_signals = eng.variational_mode_decomposition(filepath, 1, 6)
vmd_signals_np = np.array(vmd_signals)
print("VMD shape:", vmd_signals_np.shape)
```

```python
# Extract features
df_original = extract_from_signals(signal, "Original")
df_emd = extract_from_signals(np.array(emd_signals), "EMD")
df_ewt = extract_from_signals(np.array(ewt_signals), "EWT")
df_vmd = extract_from_signals(np.array(vmd_signals), "VMD")

# Combine
final_features = pd.concat([df_original, df_emd, df_ewt, df_vmd])

final_features
#final_features.to_csv("EEG_all_features_channel1.csv", index=False)
```
